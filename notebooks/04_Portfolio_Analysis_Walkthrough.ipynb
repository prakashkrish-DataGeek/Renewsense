```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid')
```

### 1. Initializing the Portfolio Scores Matrix
We define a three-site portfolio for evaluation under multiple scenarios:
1. **Rajasthan (Solar asset focus)**
2. **Northern Germany (Wind asset focus)**
3. **Atacama Desert (High solar yield, poor infrastructure)**

```python
portfolio_scores = {
    'Rajasthan Solar': {'SRI': 85.0, 'WRP': 40.0, 'HSI': 20.0, 'GIR': 80.0, 'EVI': 85.0, 'RPE': 78.0, 'LSA': 75.0},
    'Germany Wind': {'SRI': 35.0, 'WRP': 88.0, 'HSI': 85.0, 'GIR': 90.0, 'EVI': 95.0, 'RPE': 94.0, 'LSA': 45.0},
    'Atacama Desert': {'SRI': 98.0, 'WRP': 68.0, 'HSI': 10.0, 'GIR': 40.0, 'EVI': 70.0, 'RPE': 82.0, 'LSA': 90.0}
}
df_portfolio = pd.DataFrame(portfolio_scores)
df_portfolio
```

### 2. Scenario Weighting Overrides
We evaluate all three sites under two contrasting weighting profiles:
- **Conservative Infrastructure**: Heavy weights on grid (0.25) and regulatory (0.25).
- **Climate-First**: Heavy weights on resource yield (solar 0.20, wind 0.20) and water basin stability (0.20).

```python
scenarios = {
    'Conservative Infrastructure': {'SRI': 0.10, 'WRP': 0.10, 'HSI': 0.10, 'GIR': 0.25, 'EVI': 0.10, 'RPE': 0.25, 'LSA': 0.10},
    'Climate-First': {'SRI': 0.20, 'WRP': 0.20, 'HSI': 0.20, 'GIR': 0.10, 'EVI': 0.10, 'RPE': 0.10, 'LSA': 0.10}
}

def calculate_geometric_iss(scores, weights):
    return np.prod([s ** w for s, w in zip(scores, weights)])

results = []
for site, v_scores in portfolio_scores.items():
    scores_list = [v_scores[k] for k in ['SRI', 'WRP', 'HSI', 'GIR', 'EVI', 'RPE', 'LSA']]
    row = {'Site Name': site}
    for scen_name, w_dict in scenarios.items():
        w_list = [w_dict[k] for k in ['SRI', 'WRP', 'HSI', 'GIR', 'EVI', 'RPE', 'LSA']]
        row[scen_name] = round(calculate_geometric_iss(scores_list, w_list), 1)
    results.append(row)
    
df_comparison = pd.DataFrame(results)
df_comparison
```

### 3. Insights & Discussion
- Under **Climate-First**, Atacama Desert is highly competitive due to exceptional solar GHI yield.
- Under **Conservative Infrastructure**, Germany Wind and Rajasthan outperform Atacama significantly due to secure grid interconnections (GIR) and favorable regulatory permitting timelines.
