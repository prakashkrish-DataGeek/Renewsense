**Description:** Exploratory Data Analysis comparing Global Horizontal Irradiance (GHI) and Wind Power Density (WPD) globally to map complementary clean energy asset co-locations.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set aesthetic styling matching RenewSense theme
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = [10, 6]
plt.rcParams['font.sans-serif'] = 'Source Sans Pro'
```

### 1. Simulating Global Resource Distributions
We generate 1,000 global coordinates representing solar and wind potentials to analyze complementary characteristics.

```python
np.random.seed(42)
n_points = 1000

# GHI ranges from 1.0 (polar) to 7.8 (tropics/deserts)
ghi = np.random.normal(loc=4.5, scale=1.5, size=n_points)
ghi = np.clip(ghi, 1.0, 8.0)

# Wind Power Density ranges from 50 to 1000 W/m²
# Often, wind potential is inversely proportional to peak solar zones (e.g. desert calms)
wpd = 1200 - (ghi * 120) + np.random.normal(loc=0, scale=150, size=n_points)
wpd = np.clip(wpd, 50, 1000)

df = pd.DataFrame({
    'GHI_kWh_m2_day': ghi,
    'Wind_PD_Wm2': wpd
})
df.head()
```

### 2. Scatter Plot: Solar vs Wind Complementarities
We investigate whether high solar resource zones suffer from poor wind potential, and locate the sweet spots for dual-resource co-location.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='GHI_kWh_m2_day', y='Wind_PD_Wm2', alpha=0.6, color='#1B6CA8')
plt.title('Global Solar (GHI) vs Wind Power Density (WPD) Distributions', fontsize=14, color='#1B6CA8')
plt.xlabel('Global Horizontal Irradiance (kWh/m²/day)')
plt.ylabel('Wind Power Density (W/m²)')
plt.axvline(x=5.5, color='#27AE60', linestyle='--', label='Favourable Solar (>5.5)')
plt.axhline(y=400, color='#17A589', linestyle='--', label='Favourable Wind (>400 W/m²)')
plt.legend()
plt.show()
```
